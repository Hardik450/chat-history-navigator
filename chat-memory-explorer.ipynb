```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os
load_dotenv()
google_api_key = os.getenv("GOOGLE_API_KEY")
model = 'gemini-2.0-flash'
chat = ChatGoogleGenerativeAI(model = model, temperature = 0.7)
```

```python
from langchain.memory import ConversationBufferMemory, ConversationBufferWindowMemory, ConversationTokenBufferMemory, ConversationSummaryMemory

# memory = ConversationBufferMemory()
# memory = ConversationBufferWindowMemory(k=2)
# memory = ConversationTokenBufferMemory(llm=chat, max_token_limit=1000)
memory = ConversationSummaryMemory(llm=chat, max_token_limit=1000)
```

```python
from langchain.chains import ConversationChain

conversation = ConversationChain(
    llm = chat,
    memory = memory,
    verbose = True,
)
```

```python
print("\n🤖 Chatbot is ready! Type 'exit' to end the chat.")
print("====================================================")

while True:
    user_input = input("\nYou: ")
    if user_input.lower() in ['exit', 'quit', 'bye', 'goodbye']:
        print("\n👋 Chatbot says: Goodbye! It was nice chatting with you")
        break

    response = conversation.invoke(user_input)
    print(f"\n🤖 Chatbot says: {response}")
    
```
